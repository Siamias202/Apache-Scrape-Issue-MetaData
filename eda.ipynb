```python
import requests
import xml.etree.ElementTree as ET
from tqdm import tqdm

def get_resolved(product, issue_id):
    issue_key = f"{product}-{issue_id}"

    try:
        url = (
            f"https://issues.apache.org/jira/si/jira.issueviews:issue-xml/"
            f"{issue_key}/{issue_key}.xml"
        )

        response = requests.get(url, timeout=30)
        response.raise_for_status()

        root = ET.fromstring(response.content)
        item = root.find("channel/item")

        if item is None:
            return None

        return item.findtext("resolved")

    except Exception:
        return None


df["resolved"] = [
    get_resolved(prod, issue_id)
    for prod, issue_id in tqdm(
        zip(df["product"], df["issue_id"]),
        total=len(df)
    )
]
```

```python
import requests
import xml.etree.ElementTree as ET

issue_key = "ZOOKEEPER-66"

url = f"https://issues.apache.org/jira/si/jira.issueviews:issue-xml/{issue_key}/{issue_key}.xml"

response = requests.get(url, timeout=30)
response.raise_for_status()

xml_data = response.content

root = ET.fromstring(xml_data)

item = root.find("channel/item")

data = {
    "key": item.findtext("key"),
    "summary": item.findtext("summary"),
    "status": item.findtext("status"),
    "priority": item.findtext("priority"),
    "reporter": item.findtext("reporter"),
    "assignee": item.findtext("assignee"),
    "created": item.findtext("created"),
    "updated": item.findtext("updated"),
    "resolved": item.findtext("resolved"),
    "description": item.findtext("description"),
}

print(data)
```
